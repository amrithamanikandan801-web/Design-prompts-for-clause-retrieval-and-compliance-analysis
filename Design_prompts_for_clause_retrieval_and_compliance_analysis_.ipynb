```python
# Sample shipment issue from our shipping dataset

shipment_issue = """
The shipment was delivered 3 days later than the planned delivery date.
The vendor did not notify the buyer about the delay within 24 hours.
"""

print("Shipment Issue:")
print(shipment_issue)
```

```python
# Contract clause knowledge base

contract_clauses = [
    {
        "category": "Delivery",
        "clause": "Supplier must deliver goods within 7 business days."
    },
    {
        "category": "Delay Notification",
        "clause": "Supplier must notify the buyer within 24 hours of any delivery delay."
    },
    {
        "category": "Insurance",
        "clause": "Supplier must maintain valid transportation insurance."
    },
    {
        "category": "Payment",
        "clause": "Payment will be completed within 30 days after successful delivery."
    },
    {
        "category": "Termination",
        "clause": "Either party must provide 30 days notice before termination."
    }
]

print("Total contract clauses:", len(contract_clauses))

for clause in contract_clauses:
    print(f"\nCategory: {clause['category']}")
    print(f"Clause: {clause['clause']}")
```

```python
def retrieve_relevant_clauses(shipment_issue, contract_clauses):
    """
    Retrieve contract clauses relevant to a shipment issue.
    """

    keywords = {
        "delivery": ["delivery", "delivered", "days", "late", "delay"],
        "delay notification": ["delay", "notify", "notification", "24 hours"],
        "insurance": ["insurance", "insured"],
        "payment": ["payment", "paid"],
        "termination": ["termination", "terminate", "notice"]
    }

    issue_text = shipment_issue.lower()

    matched_clauses = []

    for clause in contract_clauses:
        category = clause["category"].lower()
        clause_text = clause["clause"].lower()

        search_words = keywords.get(category, [])

        if any(word in issue_text for word in search_words):
            matched_clauses.append(clause)

    return matched_clauses


# Retrieve relevant clauses
relevant_clauses = retrieve_relevant_clauses(
    shipment_issue,
    contract_clauses
)

print("Relevant Clauses Found:")
print("-" * 50)

for clause in relevant_clauses:
    print("Category:", clause["category"])
    print("Clause:", clause["clause"])
    print()
```

```python
def analyze_compliance(shipment_issue, relevant_clauses):
    """
    Analyze shipment compliance against relevant contract clauses.
    """

    issue_text = shipment_issue.lower()

    results = []

    for clause in relevant_clauses:
        category = clause["category"]
        clause_text = clause["clause"]

        if category == "Delivery":
            if "late" in issue_text or "delay" in issue_text or "3 days later" in issue_text:
                status = "Non-Compliant"
                violation = "Delivery exceeded the contractual delivery period."
                risk = "High"
                recommendation = "Review vendor performance and initiate the delay procedure."
            else:
                status = "Compliant"
                violation = "No delivery violation identified."
                risk = "Low"
                recommendation = "Continue monitoring vendor performance."

        elif category == "Delay Notification":
            if "did not notify" in issue_text or "not notify" in issue_text:
                status = "Non-Compliant"
                violation = "Required delay notification was not provided within 24 hours."
                risk = "High"
                recommendation = "Request an explanation from the vendor and review the contract's delay procedure."
            else:
                status = "Compliant"
                violation = "No notification violation identified."
                risk = "Low"
                recommendation = "Continue monitoring compliance."

        else:
            status = "Needs Review"
            violation = "No automatic rule available for this clause."
            risk = "Medium"
            recommendation = "Perform manual contract review."

        results.append({
            "Category": category,
            "Clause": clause_text,
            "Status": status,
            "Violation": violation,
            "Risk": risk,
            "Recommendation": recommendation
        })

    return results


# Run compliance analysis
compliance_results = analyze_compliance(
    shipment_issue,
    relevant_clauses
)

for result in compliance_results:
    print("Category:", result["Category"])
    print("Status:", result["Status"])
    print("Violation:", result["Violation"])
    print("Risk:", result["Risk"])
    print("Recommendation:", result["Recommendation"])
    print("-" * 60)
```

```python
import pandas as pd

compliance_report = pd.DataFrame(compliance_results)

print("COMPLIANCE ANALYSIS REPORT")
print("=" * 80)

display(compliance_report)
```

```python
# Determine overall risk

if "High" in compliance_report["Risk"].values:
    overall_risk = "High"
elif "Medium" in compliance_report["Risk"].values:
    overall_risk = "Medium"
else:
    overall_risk = "Low"

print("Overall Compliance Risk:", overall_risk)
```

```python
import pandas as pd

filename = list(uploaded.keys())[0]

shipping_df = pd.read_csv(filename)

print("File loaded:", filename)
print("Dataset Shape:", shipping_df.shape)

display(shipping_df.head())


```

```python
def create_shipment_issue(row):
    if row["Delay_Days"] > 0:
        return (
            f"Vendor {row['Vendor']} delivered shipment "
            f"{row['Shipment_ID']} {row['Delay_Days']} days late. "
            f"Planned delivery was {row['Planned_Delivery_Days']} days "
            f"and actual delivery was {row['Actual_Delivery_Days']} days."
        )
    else:
        return (
            f"Vendor {row['Vendor']} delivered shipment "
            f"{row['Shipment_ID']} within the planned delivery period."
        )


shipping_df["Shipment_Issue"] = shipping_df.apply(
    create_shipment_issue,
    axis=1
)

display(
    shipping_df[
        [
            "Shipment_ID",
            "Vendor",
            "Planned_Delivery_Days",
            "Actual_Delivery_Days",
            "Delay_Days",
            "Shipment_Issue"
        ]
    ].head(10)
)
```

```python
def retrieve_clause_for_shipment(shipment_issue):
    matched = retrieve_relevant_clauses(
        shipment_issue,
        contract_clauses
    )

    if len(matched) == 0:
        return "No Relevant Clause"

    return " | ".join(
        [
            f"{c['category']}: {c['clause']}"
            for c in matched
        ]
    )


shipping_df["Relevant_Clause"] = shipping_df["Shipment_Issue"].apply(
    retrieve_clause_for_shipment
)

display(
    shipping_df[
        [
            "Shipment_ID",
            "Vendor",
            "Shipment_Issue",
            "Relevant_Clause"
        ]
    ].head(10)
)
```

```python
def analyze_shipment_compliance(row):

    planned = row["Planned_Delivery_Days"]
    actual = row["Actual_Delivery_Days"]
    delay = row["Delay_Days"]

    # Delivery compliance
    if delay > 0:
        status = "Non-Compliant"
        risk = "High"

        violation = (
            f"Shipment delayed by {delay} day(s). "
            f"Planned delivery: {planned} days; "
            f"Actual delivery: {actual} days."
        )

        recommendation = (
            "Review vendor performance and check "
            "the contract delay requirements."
        )

    else:
        status = "Compliant"
        risk = "Low"

        violation = "No delivery delay detected."

        recommendation = (
            "Continue monitoring vendor performance."
        )

    return pd.Series({
        "Compliance_Status": status,
        "Compliance_Risk": risk,
        "Compliance_Violation": violation,
        "Recommendation": recommendation
    })


shipping_df[
    [
        "Compliance_Status",
        "Compliance_Risk",
        "Compliance_Violation",
        "Recommendation"
    ]
] = shipping_df.apply(
    analyze_shipment_compliance,
    axis=1
)

print("Compliance analysis completed for all shipments.")
```

```python
compliance_columns = [
    "Shipment_ID",
    "Vendor",
    "Delay_Days",
    "Delayed",
    "Relevant_Clause",
    "Compliance_Status",
    "Compliance_Risk",
    "Compliance_Violation",
    "Recommendation"
]

display(
    shipping_df[compliance_columns].head(20)
)
```

```python
print("COMPLIANCE SUMMARY")
print("=" * 50)

print("\nCompliance Status:")
print(shipping_df["Compliance_Status"].value_counts())

print("\nRisk Distribution:")
print(shipping_df["Compliance_Risk"].value_counts())
```

```python
vendor_evaluation = (
    shipping_df
    .groupby("Vendor")
    .agg(
        Total_Shipments=("Shipment_ID", "count"),
        Delayed_Shipments=("Delay_Days", lambda x: (x > 0).sum()),
        Average_Delay_Days=("Delay_Days", "mean"),
        Compliant_Shipments=(
            "Compliance_Status",
            lambda x: (x == "Compliant").sum()
        ),
        Non_Compliant_Shipments=(
            "Compliance_Status",
            lambda x: (x == "Non-Compliant").sum()
        ),
        High_Risk_Shipments=(
            "Compliance_Risk",
            lambda x: (x == "High").sum()
        )
    )
    .reset_index()
)

vendor_evaluation["Delay_Rate_%"] = (
    vendor_evaluation["Delayed_Shipments"]
    / vendor_evaluation["Total_Shipments"]
    * 100
).round(2)

vendor_evaluation["Compliance_Rate_%"] = (
    vendor_evaluation["Compliant_Shipments"]
    / vendor_evaluation["Total_Shipments"]
    * 100
).round(2)

vendor_evaluation["Average_Delay_Days"] = (
    vendor_evaluation["Average_Delay_Days"]
    .round(2)
)

display(vendor_evaluation)
```

```python
def assign_vendor_risk(row):

    if row["Delay_Rate_%"] >= 50 or row["High_Risk_Shipments"] >= 5:
        return "High"

    elif row["Delay_Rate_%"] >= 25 or row["High_Risk_Shipments"] >= 2:
        return "Medium"

    else:
        return "Low"


vendor_evaluation["Overall_Risk"] = vendor_evaluation.apply(
    assign_vendor_risk,
    axis=1
)

display(vendor_evaluation)
```

```python
print("VENDOR RISK SUMMARY")
print("=" * 50)

print(
    vendor_evaluation["Overall_Risk"].value_counts()
)
```

```python
vendor_report = vendor_evaluation[
    [
        "Vendor",
        "Total_Shipments",
        "Delayed_Shipments",
        "Delay_Rate_%",
        "Average_Delay_Days",
        "Compliant_Shipments",
        "Non_Compliant_Shipments",
        "High_Risk_Shipments",
        "Compliance_Rate_%",
        "Overall_Risk"
    ]
].copy()

display(vendor_report)
```

```python
vendor_report["Vendor_Score"] = (
    100
    - (vendor_report["Delay_Rate_%"] * 0.6)
    - (vendor_report["High_Risk_Shipments"] /
       vendor_report["Total_Shipments"] * 100 * 0.4)
).clip(0, 100).round(2)

display(vendor_report)
```

```python
def vendor_recommendation(row):

    if row["Overall_Risk"] == "High":
        return (
            "Detailed vendor review required. "
            "Monitor delivery delays and contract compliance."
        )

    elif row["Overall_Risk"] == "Medium":
        return (
            "Monitor vendor performance and "
            "review recurring delays."
        )

    else:
        return (
            "Continue routine monitoring of "
            "delivery and compliance performance."
        )


vendor_report["Recommendation"] = vendor_report.apply(
    vendor_recommendation,
    axis=1
)

display(vendor_report)
```
